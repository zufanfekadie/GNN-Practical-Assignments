# Assignment 1: Build and Verify a GCN from Scratch

## 1. Overview and Problem Statement
In this assignment, we construct a 2-layer Graph Convolutional Network (GCN) entirely from scratch using pure PyTorch tensor operations—without ready-made GNN libraries.

### Key Requirements Addressed:
1. **Synthetic Graph**: At least 100 nodes (we use 120), exactly 3 numerical features, 2 target classes (0 and 1).
2. **Graph Structure Utility**: Designed with overlapping feature distributions (making feature-only models ambiguous) and homophilic community connections (allowing graph convolutions to resolve the ambiguity).
3. **Scratch Implementation**: Explicit self-loops ($\tilde{A} = A + I$), degree normalization ($\hat{A} = \tilde{D}^{-1/2} \tilde{A} \tilde{D}^{-1/2}$), message passing, and two-layer forward pass.
4. **Correctness Checks**:
   - Single-node output matched against explicit manual calculation.
   - Permutation equivariance verified across random node reorderings.
5. **Ablation Studies**: Rigorous comparison against an MLP baseline, randomized edges, and empty graph.
